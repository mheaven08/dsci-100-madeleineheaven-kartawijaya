Visualization 1: The visualization below displays the relationship between a team’s tournament seed and its final position at the tournament. By comparing the mean seed across different tournament results, it helps us identify whether top seeded teams (lower seed numbers) actually advance further in the tournament than lower seeded teams (higher seed numbers). This is relevant to the analysis because it allows us to examine whether seed is a strong predictor of a team’s tournament results and whether top-seeded teams are likely to reach a further stage. 

Note: Seed 0 represents a designation in the dataset for 1983 opening-round teams competing for the No. 8 seed

```r
# group ncaa_tidy by tournament finish category and calculate the mean seed for each category
seed_finish_plot <- ncaa_tidy |> group_by(tourney_finish) |> summarise(mean_seed = mean(seed)) |> 

# pipe summarized data into ggplot and order tournament finish categories by mean seed on the y-axis
ggplot(aes(x = mean_seed, y = reorder(tourney_finish, mean_seed))) + 

# add a bar chart layer using calculated mean values,
# add black outline on each bar, and fill each bar with steelblue color to make it more clear
geom_bar(stat = "identity", color = "black", fill = "steelblue") + 

# set axis labels and plot title
labs(x = "Mean Team Seed", y = "Tournament Finish", title = "Mean Team Seed by Tournament Finish") + 

# set text size to 16 so it is easier to read
theme(text = element_text(size = 16)) + 

# relabel y-axis categories to make them human-readable
scale_y_discrete(labels = c(
    "first_round_loss" = "First Round Loss", 
    "second_round_loss" = "Second Round Loss",
    "top_16_loss" = "Top 16 Loss",
    "top_8_loss" = "Top 8 Loss",
    "top_4_loss" = "Top 4 Loss",
    "top_2_loss" = "Top 2 Loss",
    "opening_round_loss" = "Opening Round Loss",
    "champ" = "Champion"))

# display the mean seed by tournament finish bar chart
seed_finish_plot
```

Visualization 2: The visualization below displays the relationship between a team’s regular-season winning percentage and its final tournament finish. By comparing the mean winning percentage across different tournament results, it helps identify whether teams with higher winning percentages actually advance further in the tournament than teams with lower winning percentages. This is relevant to the analysis because it allows us to examine whether regular-season winning percentage is a strong predictor (reliably leads to tournament success) of a team’s tournament results and whether teams with higher winning percentages are likely to achieve better results.

```r
# group ncaa_tidy by tournament finish category and calculate the mean regular season winning percentage for each category
wins_finish_plot <- ncaa_tidy |> group_by(tourney_finish) |> summarise(mean_wins = mean(reg_wins_pct)) |> 

# pipe summarized data into ggplot and order tournament finish categories by mean reular season winning percentage on the y-axis
ggplot(aes(x = mean_wins, y = reorder(tourney_finish, mean_wins))) + 

# add a bar chart layer using calculated mean values,
# add black outline on each bar, and fill each bar with steelblue color to make it more clear
geom_bar(stat = "identity", color = "black", fill = "steelblue") + 

# set axis labels and plot title
labs(x = "Mean Regular Season Winning Percentage", y = "Tournament Finish", title = "Mean Reg. Season Win % by Tourney Finish") + 

# set text size to 15.5 so it is easier to read
theme(text = element_text(size = 15.5)) + 

# relabel y-axis categories to make them human-readable
scale_y_discrete(labels = c(
    "first_round_loss" = "First Round Loss", 
    "second_round_loss" = "Second Round Loss",
    "top_16_loss" = "Top 16 Loss",
    "top_8_loss" = "Top 8 Loss",
    "top_4_loss" = "Top 4 Loss",
    "top_2_loss" = "Top 2 Loss",
    "opening_round_loss" = "Opening Round Loss",
    "champ" = "Champion"))

# display the mean regular season winning percentage by tournament finish bar chart
wins_finish_plot
```

To answer my research question, I will use a k-nn (k nearest neighbor) classification because my goal is to predict a categorical variable (tourney_finish) by comparing it with previous teams that have similar seed and regular season winning percentage. Potential limitations of this method are its performance depends on the value of k, and sometimes tournament outcomes can be unpredictable (similar performance in the regular season does not guarantee similar results in the tournament). To apply the model, I will conduct an initial split, applying tourney_finish as a strata, on 75% of the data as training data and the remaining 25% as testing data. After that, I will standardize the predictors (seed and reg_wins_pct) using the training data to ensure that both predictors are on the same scale, preventing one predictor from having a greater influence because of its numerical range. I will then perform a 5 fold cross-validation on the training set to select the optimal k. To assess the performance of the model, I will use accuracy to examine the proportion of correctly classified tournament finishes, and confusion matrix to examine which tournament finish categories the model predicts correctly. It is essential to use both accuracy and confusion matrix since a high accuracy model does not necessarily mean it performs well across all categories (it might predict most common category accurately but struggle to classify rare categories), by using confusion matrix, we can identify the weaknesses by showing which categories are correctly classified and which are misclassified.
